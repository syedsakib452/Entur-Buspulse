```python
table = pq.read_table("gtfs_shapes.parquet")
print(table.schema)
df = table.to_pandas()

print("meta data for gtfs_shapes.parquet:")
print(pq.ParquetFile("gtfs_shapes.parquet").metadata)

print(df.iloc[0].shape_id)      # first row, as a column/value listing
```

A shape is useless on it's own but add stop times and trips it tells what each shape is supposed to travel to.
So trips.txt connects a shape to a route (via route_id) and to a trip, and stop_times then connects that trip to its stops. In your example, KOL:JourneyPattern:1001_Inbound_... is just a shape ID. The KOL prefix is the data provider’s codespace, and Inbound is the direction, but nothing more can be read from it.

```python
import pandas as pd

shape_id = "KOL:JourneyPattern:1001_Inbound_251008118370718_250903083108828"

trips = pd.read_parquet("gtfs_trips.parquet")
routes = pd.read_parquet("gtfs_routes.parquet")
stop_times = pd.read_parquet("gtfs_stop_times.parquet", columns=["trip_id", "stop_id", "stop_sequence"])
stops = pd.read_parquet("gtfs_stops.parquet", columns=["stop_id", "stop_name"])
print(pq.ParquetFile("gtfs_stop_times.parquet").metadata)
print(pq.ParquetFile("gtfs_stops.parquet").metadata)


# route(s) using this shape
t = trips[trips["shape_id"] == shape_id]
print(t[["route_id", "trip_id"]].head())
print(routes[routes["route_id"].isin(t["route_id"])][["route_id", "route_short_name", "route_long_name", "route_type"]])

# stops in order, using one trip that follows this shape
one_trip = t["trip_id"].iloc[0]
s = stop_times[stop_times["trip_id"] == one_trip].copy()
s["stop_sequence"] = s["stop_sequence"].astype(int)
print(s.sort_values("stop_sequence").merge(stops, on="stop_id")[["stop_sequence", "stop_name"]])
```
